## Example 5.1.1

```julia
using Plots
theme(:dark); default(thickness_scaling=1.5)
n = 5
t = range(-1, 1, n+1)
y = @. t^2 + t + 0.05 * sin(20t)
scatter(t, y, legend=false)
```

```julia
using Polynomials
p = Polynomials.fit(t, y, n)     # interpolating polynomial
plot!(p, -1, 1)
```

```julia
n = 18
t = range(-1, 1, n+1)
y = @. t^2 + t + 0.05 * sin(20t)
scatter(t, y, leg=false)
```

```julia
p = Polynomials.fit(t, y, n)
x = range(-1, 1, 1000)    # use a lot of points
plot!(p, -1, 1)
```

## Example 5.1.3

```julia
using Plots
theme(:dark); default(thickness_scaling=1.5)
n = 12
t = range(-1, 1, n+1)
y = @. t^2 + t + 0.5 * sin(20t)
scatter(t, y, label="", leg=:top)
```

```julia
using FNCFunctions
p = FNC.plinterp(t, y)
plot!(p, -1, 1, label="piecewise linear")
```

```julia
using Dierckx
p = Spline1D(t, y)
plot!(x -> p(x), -1, 1, label="piecewise cubic")
```

## Example 5.1.4

```julia
using Plots, LaTeXStrings
theme(:dark); default(thickness_scaling=1.5)

n = 18
t = range(-1, 1, n+1)
y = [zeros(9); 1; zeros(n - 9)];  # data for 10th cardinal function

scatter(t, y, label=latexstring("cardinal for \$t_9\$"))
```

```julia
using Dierckx
ϕ = Spline1D(t, y)
plot!(x -> ϕ(x), -1, 1; label="spline", xlabel=L"x",  ylabel=L"\phi(x)")
```

```julia
using Polynomials

scatter(t, y, label="data")

ϕ = Polynomials.fit(t, y, n)
plot!(x -> ϕ(x), -1, 1; label="polynomial",  legend=:top, xlabel=L"x",  ylabel=L"\phi(x)")
```
