## Example 5.4.5

$f(x) = \cos(x^2)$, find $f'(0.5)$

````julia
t = [0.35, 0.5, 0.57, 0.6, 0.75]   # nodes
f(x) = cos(x^2)
df_dx(x) = -2 * x * sin(x^2)
exact_value = df_dx(0.5)
````

````julia
using FNCFunctions
t̂ = t .- 0.5
w = FNC.fdweights(t̂, 1)
````

````julia
using LinearAlgebra
fd_value = dot(w, f.(t))
````
